Изучите сортировку вставками и реализуйте ее в виде функции insertion_sort(some_list). Функция принимает список и сортирует его по возрастанию.


```python
def insertion_sort(some_list):
    for i in range (1, len(some_list)):
        x = some_list[i]
        j = i
        while j > 0 and some_list[j - 1] > x:
            some_list[j] = some_list[j - 1]
            j -= 1
        some_list[j] = x
    return some_list

test_list = [5, 3, 2, 54, 5, 4, 1, 46, 95, 8, 4]
print(insertion_sort(test_list))
```

b.	почти отсортированный в обратную сторону список (один элемент в середине списка находится не на своем месте); 

```python
def list_m_reversed(n):
    arr = list(range(n, 0, -1))
    arr[n // 2], arr[n // 2 + 1] = arr[n // 2 + 1], arr[n // 2]
    return arr

sizes = [100, 500, 1000, 2000, 4000, 6000, 8000, 10000]
results_ins_revers = []

for n in sizes:
    arr = list_m_reversed(n)
    start = time.perf_counter()
    insertion_sort(arr)
    end = time.perf_counter()
    results_ins_revers.append(end - start)

plt.plot(sizes, results_ins_revers, marker='o')
plt.title("Зависимость времени от размера n")
plt.xlabel("n")
plt.ylabel("Секунд")
plt.grid(True)
plt.show()
```

c.	половина данных равна 0, половина оставшихся 1, половина оставшихся 2 и т.д.;

```python
def list_seq(n):
    arr = []
    val = 0
    left = n
    while left > 0:
        count = left // 2
        if count == 0:
            count = 1
        arr.extend([val] * count)
        left -= count
        val += 1
    return arr

sizes = [100, 500, 1000, 2000, 4000, 6000, 8000, 10000]
results_ins_seq = []

for n in sizes:
    arr = list_seq(n)
    start = time.perf_counter()
    insertion_sort(arr)
    end = time.perf_counter()
    results_ins_seq.append(end - start)

plt.plot(sizes, results_ins_seq, marker='o')
plt.title("Зависимость времени от размера n")
plt.xlabel("n")
plt.ylabel("Секунд")
plt.grid(True)
plt.show()
```

```python
def list_near_positoned(n):
    arr = list(range(n))

    for i in range(n):
        j = random.randint(max(0, i - 10), min(n - 1, i + 10))
        arr[i], arr[j] = arr[j], arr[i]
    return arr   

sizes = [100, 500, 1000, 2000, 4000, 6000, 8000, 10000]
results_ins_pos = []

for n in sizes:
    arr = list_near_positoned(n)
    start = time.perf_counter()
    insertion_sort(arr)
    end = time.perf_counter()
    results_ins_pos.append(end - start)

plt.plot(sizes, results_ins_pos, marker='o')
plt.title("Зависимость времени от размера n")
plt.xlabel("n")
plt.ylabel("Секунд")
plt.grid(True)
plt.show()
```

5.	(6 баллов) Модифицируйте сортировку вставками, применив двоичный поиск места вставки нового элемента. Реализуйте модифицированный алгоритм в виде функции insertion_sort_with_bin_search(some_list). Функция принимает список и сортирует его по возрастанию.
Для этой функции выполните задание из пункта 4.
Есть ли отличие в результатах работы функций insertion_sort и insertion_sort_with_bin_search? Как вы можете это объяснить?  


```python
def pos_bin_search(arr, x, left, right):
    while left < right:
        mid = (left + right) // 2
        if arr[mid] <= x:
            left = mid + 1
        else:
            right = mid
    return left

def insertion_sort_with_bin_search(some_list):
    for i in range(1, len(some_list)):
        x = some_list[i]
        pos = pos_bin_search(some_list, x, 0, i)
        j = i
        while j > pos:
            some_list[j] = some_list[j - 1]
            j -= 1
        some_list[pos] = x
    return some_list

test_list = [7, 3, 9, 4, 2, 5, 6, 1, 8]

print(insertion_sort_with_bin_search(test_list))
```
