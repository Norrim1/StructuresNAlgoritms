```python
import pandas as pd
df = pd.read_csv('14water.txt', sep = '\t')
df.head()
```

Используя метод .describe() вычислите описательные статистики для северных и
южных городов. Сравните средние значения смертности в северных и южных
городах и значения концентрации кальция в питьевой воде.


```python
print(df.loc[df.locatio == "Sout"].describe())
print(df.loc[df.locatio == "Nort"].describe()) #Смертность выше на севере, концентрация выше на юге
```

Постройте 95% доверительные интервалы для средней годовой смертности по
всем южным и северным городам. Отличаются ли границы интервалов?
Постройте 95% доверительные интервалы для средней концентрации кальция в
питьевой воде для южных и северных городов. Отличаются ли границы
интервалов? 

```python
from statsmodels.stats.weightstats import _tconfint_generic
import numpy as np
mort_mean = df.mortalit.mean()
mort_mean_std = df.mortalit.std() / np.sqrt(df.mortalit.shape[0])
print('95% interval:', _tconfint_generic(mort_mean, mort_mean_std, df.mortalit.shape[0] - 1, 0.05, 'two-sided'))
```
