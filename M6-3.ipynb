Обучите модель линейной регрессии на данных согласно вашему варианту. Выведите угловой коэффициент, точку пересечения с осью ординат и точность. Y = 3*x + random(5)

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

np.random.seed(42)
x = np.linspace(0, 10, 100).reshape(-1, 1) #Без reshape выдаёт ошибку
y = 3 * x.flatten() + np.random.uniform(0, 5, x.shape[0])

model = LinearRegression()
model.fit(x, y)

angle = model.coef_[0]
intercept = model.intercept_
accuracy = r2_score(y, model.predict(x))

print(angle)
print(intercept)
print(accuracy)
```
